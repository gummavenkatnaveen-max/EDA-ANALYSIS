```python
df.drop_duplicates(inplace=True)
```

```python
df['Category']=df['Category'].replace ('electronics','Electronics')
```

```python
df['Category']=df['Category'].replace ('Clothng','Clothing')
```

```python
df['Category']=df['Category'].replace (' ELECTRONICS ','Electronics')
```

```python
df['Payment_Method']=df['Payment_Method'].replace('UPI','upi')
```

```python
df['Payment_Method']=df['Payment_Method'].replace('Credit Card','Credit card')
```

```python
df['Category'].unique()
```

```python
df['Payment_Method'].unique()
```

```python
df['Customer_Segment']=df['Customer_Segment'].fillna(df['Customer_Segment'].mode()[0])
```

```python
df['Unit_Price']=df['Unit_Price'].fillna(df['Unit_Price'].median())
```

```python
df['Discount_Percent']=df['Discount_Percent'].fillna(df['Discount_Percent'].median())
```

```python
df['Revenue']=df['Revenue'].fillna(df['Revenue'].median())
```

```python
df['Cost']=df['Cost'].fillna(df['Cost'].median())
```

```python
df['Profit']=df['Profit'].fillna(df['Profit'].median())
```

```python
df['Payment_Method']=df['Payment_Method'].fillna(df['Payment_Method'].mode()[0])
```

```python
df['City'] = df.groupby('Region')['City'].transform(
    lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'Unknown')
)
```
